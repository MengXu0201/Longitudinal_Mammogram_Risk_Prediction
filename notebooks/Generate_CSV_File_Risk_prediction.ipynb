```python
import pandas as pd
from datetime import datetime
import os
```

```python
CSV_FILE_PATH = "/mnt/cv_data/users/mengxu/Longitudinal_Mammogram_Alignment/output_csv"

positive_path = os.path.join(CSV_FILE_PATH, "POSITIVE_GROUP_FINAL.csv")
negative_path = os.path.join(CSV_FILE_PATH, "NEGATIVE_GROUP_FINAL.csv")

# Load CSV files
positive_cases = pd.read_csv(
    positive_path, parse_dates=["study_date_anon", "diagnosed_date_anon"]
)
negative_cases = pd.read_csv(
    negative_path, parse_dates=["study_date_anon"]
)
```

```python
def get_stats(df, suffix=None):
    """Provides a quick summary of a dataframe."""
    try:
        print(f"DF shape: {df.shape}")
        print(f"# Patients: {df.patient_id.nunique()}")
        print(f"# Cases: {df.exam_id.nunique()}\n")
        print(f"# Images: {df.file_path_dcm.nunique()}\n")
    except Exception as e:
        print(e)
```

```python
# check status before processing

display(negative_cases)
get_stats(negative_cases)
```

```python
# check status before processing

display(positive_cases)
get_stats(positive_cases)
```

```python
# For positive cases
def calculate_positive_followup(row):
    # Compute difference in years between the exam date and cancer diagnosis date
    return (row["diagnosed_date_anon"] - row["study_date_anon"]).days / 365.25


positive_cases["years_last_followup"] = positive_cases.apply(
    calculate_positive_followup, axis=1
)
positive_cases["years_last_followup"] = positive_cases["years_last_followup"].astype(
    int
)

# check status after processing
display(positive_cases)
get_stats(positive_cases)
```

```python
# For negative cases
def calculate_negative_followup(patient_df):
    # Determine the latest exam date for each patient
    latest_date = patient_df["study_date_anon"].max()
    # Compute difference in years between the exam date and the latest available date
    patient_df["years_last_followup"] = (
        latest_date - patient_df["study_date_anon"]
    ).dt.days / 365.25
    patient_df["years_last_followup"] = patient_df["years_last_followup"].astype(int)
    return patient_df

negative_cases = negative_cases.groupby("patient_id").apply(calculate_negative_followup)


# check status after processing
display(negative_cases)
get_stats(negative_cases)
```

```python
# Combine datasets if needed
all_cases = pd.concat([positive_cases, negative_cases], ignore_index=True)

# check status after combination
display(all_cases)
get_stats(all_cases)
```

```python
# Save to CSV

OUTPUT_ROOT = "/mnt/cv_data/users/mengxu/Longitudinal_Mammogram_Alignment/output_csv"
os.makedirs(OUTPUT_ROOT, exist_ok=True)

all_cases.to_csv(os.path.join(OUTPUT_ROOT, "EMBED_combined_cases_with_followup.csv"), index=False)
```

### The following code is added by Meng to split the dataset into train/val/test sets 

```python
import pandas as pd
from datetime import datetime
import os
from sklearn.model_selection import train_test_split

df = pd.read_csv("/mnt/cv_data/users/mengxu/Longitudinal_Mammogram_Alignment/output_csv/EMBED_combined_cases_with_followup.csv")

# Make sure date column is datetime
df["study_date_anon"] = pd.to_datetime(df["study_date_anon"], errors="coerce")
```

```python
# ============================================================
# 1. Build patient-level summary for stratification
#    Use patient-level positive / negative status
# ============================================================

# Build exam-level temporary label
temp_df = df[["patient_id", "Label"]].copy()
temp_df["is_positive"] = temp_df["Label"].notna().astype(int)

# Build patient-level summary
patient_df = (
    temp_df.groupby("patient_id", as_index=False)
    .agg(
        is_positive_patient=("is_positive", "max")
    )
    .copy()
)

# ============================================================
# 2. Show patient-level positive / negative counts
# ============================================================
n_positive = (patient_df["is_positive_patient"] == 1).sum()
n_negative = (patient_df["is_positive_patient"] == 0).sum()
n_total = len(patient_df)

print("Patient-level counts:")
print(f"Total patients   : {n_total}")
print(f"Positive patients: {n_positive}")
print(f"Negative patients: {n_negative}")
print()

print("Patient-level positive / negative distribution:")
print(patient_df["is_positive_patient"].value_counts().sort_index())
print()

# ============================================================
# 3. Patient-level stratified split
#    train : val : test = 0.5 : 0.2 : 0.3
# ============================================================
patients = patient_df["patient_id"].values
patient_strata = patient_df.set_index("patient_id")["is_positive_patient"]

train_patients, temp_patients = train_test_split(
    patients,
    test_size=0.5,
    random_state=42,
    stratify=patient_strata.loc[patients]
)

val_patients, test_patients = train_test_split(
    temp_patients,
    test_size=0.6,   # temp = 0.5 of all patients, so val = 0.2 and test = 0.3
    random_state=42,
    stratify=patient_strata.loc[temp_patients]
)

# ============================================================
# 4. Add split_group to original dataframe
# ============================================================
df["split_group"] = None

df.loc[df["patient_id"].isin(train_patients), "split_group"] = "train"
df.loc[df["patient_id"].isin(val_patients), "split_group"] = "val"
df.loc[df["patient_id"].isin(test_patients), "split_group"] = "test"

# Safety check
assert df["split_group"].isna().sum() == 0, "Some rows were not assigned to any split."

# Leakage check
patient_split_check = df.groupby("patient_id")["split_group"].nunique()
assert (patient_split_check > 1).sum() == 0, "Some patients appear in more than one split."

# ============================================================
# 5. Check patient-level distribution after split
# ============================================================
patient_split_df = patient_df.copy()
patient_split_df["split_group"] = None
patient_split_df.loc[patient_split_df["patient_id"].isin(train_patients), "split_group"] = "train"
patient_split_df.loc[patient_split_df["patient_id"].isin(val_patients), "split_group"] = "val"
patient_split_df.loc[patient_split_df["patient_id"].isin(test_patients), "split_group"] = "test"

print("Patient counts by split:")
print(patient_split_df["split_group"].value_counts())
print()

patient_split_df["cancer_status"] = patient_split_df["is_positive_patient"].apply(
    lambda x: "positive" if x == 1 else "negative"
)

print("Positive/negative patient distribution by split:")
print(
    patient_split_df.groupby(["split_group", "cancer_status"]).size().unstack(fill_value=0)
)
print()

print("Positive patient ratio by split:")
print(patient_split_df.groupby("split_group")["is_positive_patient"].mean())
print()
```

```python
# ============================================================
# Check split distribution without modifying the original df
# ============================================================

df_check = df.copy()

# ------------------------------------------------------------
# 0. Basic preparation
# ------------------------------------------------------------
# positive patient:
#   at least one exam has non-empty Label
# negative patient:
#   all exams have empty Label
df_check["is_positive"] = df_check["Label"].notna().astype(int)
df_check["is_cancer_exam"] = (df_check["Label"] == "Cancer").astype(int)

# patient-level label table
patient_label_df = (
    df_check.groupby("patient_id", as_index=False)
    .agg(is_positive_patient=("is_positive", "max"))
)

patient_split_df = df_check[["patient_id", "split_group"]].drop_duplicates()

# safety check: each patient should appear in only one split
patient_split_check = patient_split_df.groupby("patient_id")["split_group"].nunique()
assert (patient_split_check > 1).sum() == 0, "Some patients appear in more than one split."

patient_label_df = patient_label_df.merge(patient_split_df, on="patient_id", how="left")

# ============================================================
# 1. Basic: split ratio + positive / negative
# ============================================================
print("============================================================")
print("1. Basic: split ratio + positive / negative")
print("============================================================")

print("\n=== Patient-level counts ===")
patient_counts = df_check.groupby("split_group")["patient_id"].nunique()
print(patient_counts)

print("\n=== Patient-level split ratio ===")
print((patient_counts / patient_counts.sum()).round(4))

print("\n=== Exam-level counts ===")
exam_counts = df_check["split_group"].value_counts().sort_index()
print(exam_counts)

print("\n=== Exam-level split ratio ===")
print((exam_counts / exam_counts.sum()).round(4))

patient_label_df["patient_status"] = patient_label_df["is_positive_patient"].map(
    {1: "positive", 0: "negative"}
)

print("\n=== Patient-level positive / negative counts ===")
print(
    patient_label_df.groupby(["split_group", "patient_status"])
    .size()
    .unstack(fill_value=0)
)

print("\n=== Patient-level positive ratio ===")
print(patient_label_df.groupby("split_group")["is_positive_patient"].mean().round(4))

df_check["exam_status"] = df_check["is_positive"].map({1: "positive", 0: "negative"})

print("\n=== Exam-level positive / negative counts ===")
print(
    df_check.groupby(["split_group", "exam_status"])
    .size()
    .unstack(fill_value=0)
)

print("\n=== Exam-level positive ratio ===")
print(df_check.groupby("split_group")["is_positive"].mean().round(4))

print("\n=== Exam-level Cancer ratio ===")
print(df_check.groupby("split_group")["is_cancer_exam"].mean().round(4))

# ============================================================
# 2. Time-to-Cancer distribution
# ============================================================
print("\n============================================================")
print("2. Time-to-Cancer distribution")
print("============================================================")

if "Time_to_Cancer_Years" in df_check.columns:
    df_check["ttc_bin"] = df_check["Time_to_Cancer_Years"].apply(
        lambda x: "Missing_TTC" if pd.isna(x)
        else "1Y" if 0 <= x <= 1
        else "2Y" if 1 < x <= 2
        else "3Y" if 2 < x <= 3
        else "4Y" if 3 < x <= 4
        else "5Y" if 4 < x <= 5
        else "Other"
    )

    print("\n=== Time-to-Cancer distribution (counts) ===")
    print(pd.crosstab(df_check["split_group"], df_check["ttc_bin"]))

    print("\n=== Time-to-Cancer distribution (normalized by split) ===")
    print(pd.crosstab(df_check["split_group"], df_check["ttc_bin"], normalize="index").round(4))
else:
    print("Column 'Time_to_Cancer_Years' not found.")

# ============================================================
# 3. Horizon coverage
# ============================================================
print("\n============================================================")
print("3. Horizon coverage")
print("============================================================")

if "Time_to_Cancer_Years" in df_check.columns:
    for h in [1, 2, 3, 4, 5]:
        df_check[f"horizon_{h}y_available"] = (
            df_check["Time_to_Cancer_Years"].notna() &
            (df_check["Time_to_Cancer_Years"] >= h)
        ).astype(int)

    print("\n=== Horizon availability counts ===")
    horizon_count_df = pd.DataFrame({
        f"{h}Y": df_check.groupby("split_group")[f"horizon_{h}y_available"].sum()
        for h in [1, 2, 3, 4, 5]
    })
    print(horizon_count_df)

    print("\n=== Horizon availability ratio ===")
    horizon_ratio_df = pd.DataFrame({
        f"{h}Y": df_check.groupby("split_group")[f"horizon_{h}y_available"].mean()
        for h in [1, 2, 3, 4, 5]
    })
    print(horizon_ratio_df.round(4))
else:
    print("Column 'Time_to_Cancer_Years' not found.")

# ============================================================
# 4. Number of exams per patient
# ============================================================
print("\n============================================================")
print("4. Number of exams per patient")
print("============================================================")

patient_exam_counts = (
    df_check.groupby(["patient_id", "split_group"])
    .size()
    .reset_index(name="n_exam")
)

print("\n=== Exams per patient (mean) ===")
print(patient_exam_counts.groupby("split_group")["n_exam"].mean().round(4))

print("\n=== Exams per patient (std) ===")
print(patient_exam_counts.groupby("split_group")["n_exam"].std().round(4))

print("\n=== Exams per patient (median) ===")
print(patient_exam_counts.groupby("split_group")["n_exam"].median())

print("\n=== Exams per patient (min / max) ===")
print(patient_exam_counts.groupby("split_group")["n_exam"].agg(["min", "max"]))
```
